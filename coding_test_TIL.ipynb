### ✅ 모범 풀이

```python
def solution_1020_answer(grid):
    rows, cols = len(grid), len(grid[0])
    visited = [[False] * cols for _ in range(rows)]
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    largest = 0

    for row in range(rows):
        for col in range(cols):
            if grid[row][col] == 1 and not visited[row][col]:
                stack = [(row, col)]
                visited[row][col] = True   # 넣을 때 방문 처리!
                size = 0
                while stack:
                    r, c = stack.pop()
                    size += 1
                    for dr, dc in directions:
                        nr, nc = r + dr, c + dc
                        if 0 <= nr < rows and 0 <= nc < cols:
                            if grid[nr][nc] == 1 and not visited[nr][nc]:
                                visited[nr][nc] = True
                                stack.append((nr, nc))
                largest = max(largest, size)
    return largest

run_tests(solution_1020_answer, cases_1020)
```

- **핵심**: `dp[i]` = i번째 날에서 **끝나는** 구간의 최대 합 → `dp[i] = max(scores[i], dp[i-1] + scores[i])`, 전체 최댓값은 매번 갱신

- **주의**: 모두 음수일 수 있으므로 최댓값을 `0`이 아닌 `scores[0]`으로 초기화

### ✅ 모범 풀이

```python
# 1) dp 리스트 버전 — 공간 O(N)
def solution_1021_answer(scores):
    n = len(scores)
    dp = [0] * n
    dp[0] = scores[0]
    best_score = dp[0]

    for i in range(1, n):
        dp[i] = max(scores[i], dp[i - 1] + scores[i])   # 새로 시작 vs 이어 붙이기
        best_score = max(best_score, dp[i])
    return best_score

# 2) 변수 버전 — 공간 O(1) (dp[i]는 dp[i-1]만 쓰기 때문)
def solution_1021_optimized(scores):
    current = best = scores[0]
    for s in scores[1:]:
        current = max(s, current + s)
        best = max(best, current)
    return best

run_tests(solution_1021_answer, cases_1021)
run_tests(solution_1021_optimized, cases_1021)
```

### ✅ 모범 풀이

```python
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1      # 왼쪽 절반(mid 포함) 버림
        else:
            right = mid - 1     # 오른쪽 절반(mid 포함) 버림
    return -1

def solution_1022_answer(orders, queries):
    return [binary_search(orders, q) for q in queries]

run_tests(solution_1022_answer, cases_1022)
```

```python
def lower_bound(arr, x):
    left, right = 0, len(arr)          # right = len(arr): '끝 바깥'도 답이 될 수 있음
    while left < right:                # '<' : 범위가 한 점으로 모이면 종료
        mid = (left + right) // 2
        if arr[mid] < x:
            left = mid + 1
        else:
            right = mid                # mid가 답일 수 있으니 버리지 않음
    return left

def similar_1022_answer(orders, ranges):
    return [lower_bound(orders, e + 1) - lower_bound(orders, s) for s, e in ranges]

run_tests(similar_1022_answer, similar_cases_1022)
```

### ✅ 모범 풀이

```python
# 1) 인접 리스트 버전 — 반대 방향 확인이 리스트 탐색 O(k)
def solution_1023_answer(n, connections):
    graph = [[] for _ in range(n + 1)]
    for a, b in connections:
        graph[a].append(b)          # 방향 그래프: 반대 방향은 추가하지 않음

    max_count = -1                  # 전부 0개여도 1번이 선택되도록
    answer = 1
    for server in range(1, n + 1):
        count = 0
        for next_server in graph[server]:
            if server in graph[next_server]:   # 반대 방향도 있으면 양방향
                count += 1
        if count > max_count:       # '>'라서 동점이면 작은 번호 유지
            max_count = count
            answer = server
    return answer

# 2) set 버전 — 반대 방향 확인이 평균 O(1)
def solution_1023_optimized(n, connections):
    edges = set(map(tuple, connections))
    count = [0] * (n + 1)
    for a, b in connections:
        if (b, a) in edges:
            count[a] += 1
    return count.index(max(count[1:]), 1)   # 인덱스 1부터 찾은 첫 최댓값 = 가장 작은 번호

run_tests(solution_1023_answer, cases_1023)
run_tests(solution_1023_optimized, cases_1023)
```

### ✅ 모범 풀이

```python
# 1) 정렬 + 딕셔너리 — O(N log N)
def solution_1027_answer(scores):
    sorted_scores = sorted(scores, reverse=True)   # 원본 보존
    ranks = {}
    for i, score in enumerate(sorted_scores):
        if score not in ranks:                     # 처음 등장할 때만 저장
            ranks[score] = i + 1
    return [ranks[s] for s in scores]

# 2) (비교용) 단순 카운트 — O(N²): 나보다 높은 점수 개수 + 1
def solution_1027_brute(scores):
    return [sum(1 for t in scores if t > s) + 1 for s in scores]

run_tests(solution_1027_answer, cases_1027)
run_tests(solution_1027_brute, cases_1027)
```
