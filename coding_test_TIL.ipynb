```python
def run_tests(func, cases):
    passed = 0
    for i, (args, expected) in enumerate(cases, 1):
        result = func(*args)
        ok = result == expected
        passed += ok
        print(f"{'✅' if ok else '❌'} #{i}  결과: {result}  |  기대값: {expected}")
    print(f"\n{passed}/{len(cases)} 통과")
```

### ✅ 모범 풀이

```python
def solution_1020_answer(grid):
    rows, cols = len(grid), len(grid[0])
    visited = [[False] * cols for _ in range(rows)]
    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    largest = 0

    for row in range(rows):
        for col in range(cols):
            if grid[row][col] == 1 and not visited[row][col]:
                stack = [(row, col)]
                visited[row][col] = True   # 넣을 때 방문 처리!
                size = 0
                while stack:
                    r, c = stack.pop()
                    size += 1
                    for dr, dc in directions:
                        nr, nc = r + dr, c + dc
                        if 0 <= nr < rows and 0 <= nc < cols:
                            if grid[nr][nc] == 1 and not visited[nr][nc]:
                                visited[nr][nc] = True
                                stack.append((nr, nc))
                largest = max(largest, size)
    return largest

run_tests(solution_1020_answer, cases_1020)
```

# 📅 2026-10-21 | DP — 최대 연속 구간 합 (Kadane)

- **알고리즘**: 동적 계획법 (카데인 알고리즘)
- **입력/출력**: `scores`(일별 운영 점수, 음수 포함) → 연속된 날짜를 1개 이상 골랐을 때 최대 합
- **핵심**: `dp[i]` = i번째 날에서 **끝나는** 구간의 최대 합 → `dp[i] = max(scores[i], dp[i-1] + scores[i])`, 전체 최댓값은 매번 갱신
- **비유**: 눈덩이 굴리기 — 눈덩이가 녹아 마이너스가 되면 버리고 그 자리에서 새로 만든다
- **복잡도**: 시간 O(N) / 공간 O(N) → 변수만 쓰면 O(1)
- **주의**: 모두 음수일 수 있으므로 최댓값을 `0`이 아닌 `scores[0]`으로 초기화

### ✅ 모범 풀이

```python
# 1) dp 리스트 버전 — 공간 O(N)
def solution_1021_answer(scores):
    n = len(scores)
    dp = [0] * n
    dp[0] = scores[0]
    best_score = dp[0]

    for i in range(1, n):
        dp[i] = max(scores[i], dp[i - 1] + scores[i])   # 새로 시작 vs 이어 붙이기
        best_score = max(best_score, dp[i])
    return best_score

# 2) 변수 버전 — 공간 O(1) (dp[i]는 dp[i-1]만 쓰기 때문)
def solution_1021_optimized(scores):
    current = best = scores[0]
    for s in scores[1:]:
        current = max(s, current + s)
        best = max(best, current)
    return best

run_tests(solution_1021_answer, cases_1021)
run_tests(solution_1021_optimized, cases_1021)
```

### ✅ 유사 문제 모범 풀이 (직접 풀어본 뒤 확인!)

```python
def similar_1021_answer(sales):
    current = best = sales[0]
    start = best_start = best_end = 0

    for i in range(1, len(sales)):
        if sales[i] > current + sales[i]:   # 새로 시작하는 게 더 이득
            current = sales[i]
            start = i
        else:                               # 기존 구간에 이어 붙임
            current += sales[i]

        if current > best:                  # '>'라서 동점이면 먼저 나온 구간 유지
            best = current
            best_start, best_end = start, i

    return [best, best_start + 1, best_end + 1]

run_tests(similar_1021_answer, similar_cases_1021)
```
